```python
output
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")
bert_causal_lm = AutoModelForCausalLM.from_pretrained("bert-base-cased")
```

```python
generated_seqs = bert_causal_lm.generate(model_inputs)
```

```python
generated_seqs
```

```python
tokenizer.batch_decode(generated_seqs, skip_special_tokens=True)
```
